### 函数

函数接收一组参数，并返回一个（也可能是多个）值。下面是一个函数示例，它用 Python 的取模运算符检查一个数能否被 3 整除（`a % b` 返回 `a` 除以 `b` 的余数；如果余数为 0，那么 `a` 就能被 `b` 整除）：


```python
def is_divisible_by_3(x):
    return x % 3 == 0

print("6 能被 3 整除吗？{0}".format(is_divisible_by_3(6)))
print("8 能被 3 整除吗？{0}".format(is_divisible_by_3(8)))
```

你也可以让函数一次返回多个值；只需要在 `return` 语句中用逗号把它们分隔开。同样，在函数调用之后接收这些返回值时，也只需要用逗号分隔要赋值的变量。下面是一个简单示例：函数接收两个数，并返回它们的和与差。


```python
def sum_and_diff(x, y):
    return x+y, x-y

a, b = sum_and_diff(3, 5)
print(a) # 3 + 5 = 8
print(b) # 3 - 5 = -2 
```

### 对 numpy 数组切片

Numpy 使用按引用传递的语义，因此它会在已有数组上创建视图，而不会隐式复制。这对于非常大的数组尤其有用，因为复制可能很慢。


```python

x = np.array([1,2,3,4,5,6])
print(x)
```

我们可以用 `[a:b]` 对数组从 `a` 到 `b-1` 进行切片。


```python
y = x[0:4]
print(y)
```

由于切片不会复制数组，修改 `y` 也会改变 `x`。


```python
y[0] = 7
print(x)
print(y)
```

如果确实想复制 `x`，应该使用 `.copy()`。


```python
x = np.array([1,2,3,4,5,6])
y = x.copy()
y[0] = 7
print(x)
print(y)
```

## 计算机上的信号 vs. 纸笔中的信号

在 EE 120 中，我们会考虑两类信号：
- 离散时间（DT）信号：它是从整数集合（$\{..., -2, -1, 0, 1, 2, ...\}$）到实数（或复数）的函数。
- 连续时间（CT）信号：它是从实数到实数（或复数）的函数。

从数学角度看，把信号视为函数非常有用。这使我们能够对信号进行各种你已经熟悉的操作，例如让两个信号相乘，或者对它们做线性组合。

### 理解时间索引

然而，在计算机上处理信号时，我们必须面对一个关键限制：**有限内存**。我们无法存储信号在每一个整数（DT 情况）或每一个实数（CT 情况）处的取值。相反，我们把信号表示为某个固定长度 $N$ 的离散数组。因此，在信号与系统中，人们约定了如何解释信号数组表示所对应的时间索引：
- 在 DT 中，第一个值被认为对应 $n=0$，最后一个值对应 $n=N-1$，总共有 $N$ 个值。由于第一个值是你所拥有信号的开端，把它称为时间“零”是合理的。这也很方便，因为它与（大多数）编程语言中的数组索引约定相同：第一个值的索引为 0，长度为 $N$ 的数组中最后一个值的索引为 $N-1$。在这 $N$ 个值之外，信号被默认认为是隐式的零。
- 在 CT 中，信号先被*采样*，然后才进入计算机进行处理；本学期后面我们会更详细地讨论这种方法。同样，和 DT 情况一样，第一个样本被认为是时间零；但根据*采样率*，我们可以给每个样本分配一个以秒为单位的时间。例如，数字音频通常是以 44.1 kHz，也就是每秒 44100 次，对麦克风上的声压（一个 CT 信号）采样得到的。这意味着每个样本之间相隔 $1/44100$ 秒，所以我们的样本分别对应 $t=0$ 秒（第 1 个样本）、$t=1/44100$ 秒（第 2 个样本）、$t=2/44100$ 秒（第 3 个样本），依此类推，直到总共 $N$ 个样本中的最后一个样本对应 $t=(N-1)/44100$ 秒。同样，在这 $N$ 个样本之外，信号被认为是零。

在真实世界的信号处理任务中，你通常会获得关于信号来源的信息，包括它到底是真正的 DT 信号，还是 CT 信号的采样版本；如果是后者，还会给出采样率。在本实验中，我们只考虑 DT 信号，并且会给出相应的时间索引。有时，为了让你逐渐熟悉这些约定，我们会要求你把一些隐式零也包括进去。

## Q2b：单边衰减指数信号

对于所有实数 $r > 0$，我们定义速率为 $r$ 的单边衰减指数信号为

$$x(n) = e^{-rn} u(n) = \begin{cases}e^{-rn} & n = 0, 1, 2, ... \\ 0 & \text{其他情况}\end{cases}$$

同样，我们采用从 $n=0$ 开始信号的约定。具体来说，这个单边衰减指数信号被称为*右边信号*，因为向右移动时它会衰减（事实上，由于第一个非零点在 $n=0$，我们称该信号为*因果*信号）。

下面的图展示了几个不同 $r$ 值下的信号，显示范围为 $n=-5$ 到 $n=20$：

![](decayingexpos.png)

随着 $r$ 增大，信号衰减到零的速度越来越快。当 $r = 0$ 时，我们得到的是单位阶跃，它完全不会衰减。


### 处理无限长信号：截断

注意，对于所有 $r > 0$，该信号都是无限长的，这与矩形信号不同：我们可以任意选取一个 $n > 0$，此时 $x(n)$ 仍然非零（事实上它为正，因为指数函数始终为正）。

这会给我们带来问题，因为我们无法存储无限多个信号值。幸运的是，有一个简单且合理的解决方案：*截断*。就像调用 “int” 函数会让 Python 通过砍掉（或称*截断*）小数部分，把浮点数转换为整数一样（例如 `int(3.1415) = 3`，也就是去掉 `.1415`），我们也可以在某个点之后不再包含信号值，从而忽略它们。例如，在上面的图中，我们在 $n=20$ 处截断信号，因此输出中只有 21 个非零点。这样做合理的原因是：对于大多数不是特别小的速率，指数信号衰减得非常快。在上面的前两幅图中，当 $n=20$ 时，信号值看起来已经衰减到零了！对于第三幅图，如果我们想捕捉所有大于约 .001 的信号值，只需要使用更多数据点（可能大约 50-100 个）即可。


### 你的任务

在 Q2a 中，我们多次重复写代码来生成信号。这里不要这样做，而是创建一个函数来生成单边衰减指数信号。这样，每次需要生成信号时，只要调用函数即可！填写下面的 `decaying_expo` 函数，并运行其下方的单元，确认你写得正确。

你的函数应按以下顺序返回两个独立的内容：
1) 时间索引。
2) 信号值。

如果你不确定如何从 Python 函数返回多个值，请回到 Q1 查看。

**提示 1：** 函数 `np.exp` 会派上用场，因为 `np.exp(x)` 会对任意数字 $x$ 计算 $e^x$。回忆 Q1：numpy 围绕*向量化*进行了设计。你可以对 numpy 数组调用 `np.exp`（或者更一般地说，任何作用于数字的 numpy 函数），它会逐元素应用该函数！例如，`np.exp(np.array([1, .5, .25]))` 会返回一个 numpy 数组，其元素分别为 $e^1, e^{.5}, e^{.25}$。

**提示 2：** 你的函数需要处理两种情况：`n_start >= 0` 和 `n_start < 0`。建议你定义两个变量：一个用于时间索引，一个用于信号。然后根据 `n_start` 给它们赋予合适的值。


```python
def decaying_expo(rate, n_start, n_end):
    """
    返回一个右边衰减指数信号，并截断到给定索引范围。
    
    参数：
    rate    - 衰减率。
    n_start - 用于生成信号的第一个时间索引。
    n_end   - 用于生成信号的最后一个时间索引。
    
    返回：
    n       - 生成信号对应的时间索引：n_start, n_start + 1, ..., n_end。
    sig     - n 中每个输入对应的信号值。
    
    如果 n_start 不小于 n_end，则会抛出 ValueError，因为这表示要生成信号的时间索引
    范围为空。
    """
   if n_start >= n_end:
        raise ValueError("n_start must be less than n_end")

    n = np.arange(n_start, n_end + 1)

    sig = np.zeros_like(n, dtype=float) #先初始全为0

    mask = n >= 0 #mask为数组，n数组中第i个位置为0，mask上第i个位置是false，反之亦然
    sig[mask] = np.exp(-rate * n[mask])#相当于只对n大于0的部分操作（numpy的布尔索引）

    return n, sig
```

## 使用 numpy 做卷积

numpy 中用于卷积的函数是 [np.convolve](https://docs.scipy.org/doc/numpy/reference/generated/numpy.convolve.html)，它接收三个参数：
- `x`：表示信号 $x(n)$ 的 numpy 数组。
- `h`：表示信号 $h(n)$ 的 numpy 数组。
- `mode`：一个字符串，用来指定是否以及如何截断卷积结果。

它会返回 $y(n) = (x * h)(n)$。根据传入 `mode` 的不同，边界处理方式会有一些细微差异。这个参数有三个选项：`full`、`same` 和 `valid`。与其用文字解释差异，不如直接通过实际例子来看。


### `mode = "full"`

当你把 `mode` 设为 `full` 时，numpy 会计算完整卷积。如果 $x$ 是长度为 $M$ 的数组，$h$ 是长度为 $N$ 的数组，那么返回数组的长度将是 $M + N - 1$。本质上，这种模式正是你目前在课堂上见到的卷积形式。


```python
np.convolve([1/2, 1/2], [1, 1, 1], 'full')
```

```python
np.convolve([1, 2, 3], [0, 1, 0.5], 'full')
```

**这个模式有什么用？** 使用 `mode="full"` 时，你*永远*不用担心数据像其他卷积模式那样被裁剪掉，下面我们就会看到这一点！所以，如果你担心裁剪问题，这个模式总是一个安全选择。


### `mode = "same"`

这种卷积模式只保留结果的“中间”部分：它会从每一侧裁掉一些数据点，直到结果长度与两个输入中较长者相同。在实践中，这是最常用的卷积模式。由于数字信号处理任务中的大多数信号都会被零填充（原因会在后续实验中学习），卷积结果的边缘通常包含零，因此把边缘裁掉通常是可以接受的。


```python
np.convolve([1/2, 1/2], [1, 1, 1], 'same')
```

```python
np.convolve([1, 2, 3], [0, 1, 0.5], 'same')
```

注意，在两种情况下，我们都有一些数据被裁掉了！使用 `same` 时，你必须小心确保结果中不会裁掉有用数据；方法是在某个信号两侧进行零填充。对于第一个卷积，如果我们改为计算


```python
np.convolve([0, 1/2, 1/2, 0], [1, 1, 1], 'same')
```

那么，我们的所有数据都被完整保留下来了。


**这个模式有什么用？** 使用 “same” 可以让你在整个代码中处理固定大小的信号（也就是说，所有数组长度相同），因为两个长度为 $L$ 的信号在 “same” 模式下卷积后，通过裁剪会返回长度为 $L$ 的信号；而你只需要通过零填充来避免数据被裁掉的问题。听起来是不是很方便？确实如此。


### `mode = "valid"`

使用 `valid` 时，numpy 只保留两个信号完全重叠时对应的卷积部分。


```python
np.convolve([1/2, 1/2], [1, 1, 1], 'valid')
```

```python
np.convolve([1, 2, 3], [0, 1, 0.5], 'valid')
```

对于第二个信号，因为两个信号长度相同，它们只在一个时刻完全重叠。因此，输出中只有一个值。


**这个模式有什么用？** 当你想对数据进行滤波，但滤波器只有在输入完全来自你的信号、且不包含任何零填充值时才有意义，那么它非常适合。例如，如果你对股票数据应用移动平均滤波器，那么取一个“部分来自股票数据、部分来自隐式零”的平均值是没有意义的。


### 关于 `mode` 的最后说明

在实验中，为避免混淆，我们通常会告诉你要使用哪种卷积模式。实践中，“same” 模式最常用，因为你要卷积的两个信号通常会被零填充到相同长度；因此输出与两个输入长度相同，这使得为信号预分配内存更容易跟踪（因为所有内容长度都一样）。

## Q3c：得到矩形，最终章

在 Q1a 中，我们看到了两个相同长度矩形信号卷积会发生什么。那么，如果我们卷积 3 个矩形信号呢（先卷积两个，再把结果与第三个卷积）？4 个呢？5 个呢？10 个呢？100 个呢？结果可能会让你惊讶。

在这个过程中，为了确保卷积结果不会发散到无穷大，我们会把矩形信号归一化，使其总和为 1。本学期后面你会证明一个有趣事实：如果两个信号的总和都为 1，那么它们卷积后的结果总和也为 1。这里我们使用长度为 2 的矩形信号；不过如果使用更大的长度，也会得到类似结果。

本题不需要你写任何代码；只需运行单元即可。


让我们看看把 2 个、3 个和 4 个矩形信号卷积在一起会发生什么。


```python
rect = np.array([1/2, 1/2])

plt.figure(figsize=(20, 4))
plt.subplot(1, 4, 1)
x = np.convolve(rect, rect, "full")
plt.stem(x)
plt.title("2 个长度为 2 的矩形信号的卷积")

plt.subplot(1, 4, 2)
x = np.convolve(x, rect, "full")
plt.stem(x)
plt.title("3 个长度为 2 的矩形信号的卷积")

plt.subplot(1, 4, 3)
x = np.convolve(x, rect, "full")
plt.stem(x)
plt.title("4 个长度为 2 的矩形信号的卷积")

plt.subplot(1, 4, 4)
x = np.convolve(x, rect, "full")
plt.stem(x)
plt.title("5 个长度为 2 的矩形信号的卷积")
plt.show()
```

我们可以看到，结果每次都变得越来越平滑（也就是说，从底部过渡到顶部会经过越来越多的信号取值），同时卷积本身也自然会带来拉伸。也许你已经对如果不断重复这个过程会发生什么有了一些直觉。借助 matplotlib 的一个很酷的功能，我们可以反复卷积这些矩形信号，并不断更新同一张图，从而观察结果如何“演化”。运行下面的单元，把 100 个矩形信号卷积在一起，并在每次卷积后显示结果。


```python
from IPython import display

numconv = 100 # 将这里改成你想进行的卷积次数
rect = np.array([1/2, 1/2])
y = np.convolve(rect, rect, "full")
for i in range(1, numconv+1):
    y = np.convolve(y, rect, "full")
    plt.clf()
    plt.stem(y)
    plt.title("{} 个长度为 2 的矩形信号的卷积".format(i))
    display.clear_output(wait=True)
    display.display(plt.gcf())
```

随着我们不断把越来越多的矩形信号卷积在一起，最终会得到一条钟形曲线！因此，如果我们想应用大量冲激响应均为矩形的滤波器（例如移动平均滤波器），也可以等价地应用一个*高斯滤波器*，也就是冲激响应为高斯函数（钟形曲线）的滤波器。高斯滤波器在各种信号处理平滑场景中都会出现；正如你在这里看到的，它是应用大量矩形滤波器（即移动平均）的极限情况。


## 选读：从概率论角度看 Q3c

信不信由你，通过卷积大量矩形信号得到类似高斯的信号并不是巧合。它其实是概率论中最著名的结果之一——[中心极限定理](https://en.wikipedia.org/wiki/Central_limit_theorem)——的一个结果。该定理说明，在一定条件下，当你把越来越多的随机变量相加时，结果会趋近于高斯分布，也就是熟悉的钟形曲线。EECS 126 中会讲到一个有趣事实：如果你把两个独立随机变量相加，所得随机变量的分布就是这两个随机变量分布的**卷积**。因此，把 100 个矩形信号卷积在一起，实际上等价于把 100 个独立的均匀随机变量相加（均匀随机变量的概率质量函数是一个矩形，因为每个取值等可能），于是我们看到结果具有高斯分布！
